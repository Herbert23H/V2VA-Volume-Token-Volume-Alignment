```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# ====== 数据（你填自己的）======
methods = ['GAP', 'AttnPool', 'CLS', 'Token2Volume']

params = np.array([23.63, 38.11, 37.81, 27.14])      # M
flops = np.array([11151.88, 11307.41, 13009.78, 11673.38])       # G
time = np.array([531.91, 536.29, 611.48, 544.58])         # ms
auc = np.array([83.75, 84.10, 83.95, 92.77]) # %

# ====== 气泡大小（建议做归一化，不然差距太大会难看）======
size = params * 110  # 可以调这个scale

# ====== 绘图 ======
plt.figure(figsize=(8,4))

scatter = plt.scatter(
    flops,
    auc,
    s=size,
    c=time,
    alpha=0.6
)

# ====== 添加方法名称标注 ======
for i, txt in enumerate(methods):
    if txt == "Token2Volume":
        plt.text(flops[i], auc[i], txt, fontsize=10,
                 ha='center', va='center', color='red')
    else:
        plt.text(flops[i], auc[i], txt, fontsize=10,
                 ha='center', va='center', color='black')

# ====== 坐标轴 ======
plt.xlabel('FLOPs (G)')
# plt.xscale('log')
plt.ylabel('Average AUC (%)')
plt.title('Efficiency vs Performance Trade-off')

# ====== colorbar 表示推理时间 ======
cbar = plt.colorbar(scatter)
cbar.set_label('Inference Time (ms)')

# ====== 网格 ======
plt.grid(True)

plt.tight_layout()
plt.show()
```
